```python
# 1. Parse an arbitrary string and use displaCy
import spacy
from spacy import displacy
from pathlib import Path
from google.colab import files

nlp = spacy.load("en_core_web_sm")

doc = nlp("You either die a hero, or you live long enough to see yourself become the villain.")

svg = displacy.render(doc, style="dep", jupyter=False)
output_path = Path("parse.svg")
output_path.open("w", encoding="utf-8").write(svg)
print("Image saved to parse.svg")

files.download("parse.svg")
```

```python
# 2. Merged Strings operations
str1 = "Why do we fall?"
str2 = "So we can learn to pick ourselves up."

doc1 = nlp(str1)
doc2 = nlp(str2)

# Tokens
print("\nTokens in String 1:", [token.text for token in doc1])
print("Tokens in String 2:", [token.text for token in doc2])

# Text
print("\nText 1:", doc1.text)
print("Text 2:", doc2.text)

# Merge strings
merged_text = str1 + " " + str2
merged_doc = nlp(merged_text)
print("\n" + merged_text)

# Tokens and lemmas
print("\nMerged string's tokens and lemmas :")
for token in merged_doc:
    print(f"{token.text} -> {token.lemma_}")

# Span function
span = merged_doc[3:9]
print("\nTokens in the extracted span:")
for token in span:
    print(token.text)
```

```python
# 3. Jean-Claude problem
from spacy.language import Language

text = "Jean-Claude, can you help please ?, ... , I will give it you back ! Never forget ... Thank you"

doc_default = nlp(text)

print("Sentences WITHOUT customization:")
for sent in doc_default.sents:
    print("-", sent.text)

print("\nTokens and Lemmas:")
for token in doc_default:
    print(f"{token.text} -> {token.lemma_}")

@Language.component("custom_ellipsis_segmenter")
def custom_ellipsis_segmenter(doc):
    for token in doc[:-1]:
        if token.text == "...":
            doc[token.i + 1].is_sent_start = True
    return doc

nlp_custom = spacy.load("en_core_web_sm")
nlp_custom.add_pipe("custom_ellipsis_segmenter", before="parser")

doc_custom = nlp_custom(text)

print("\nSentences WITH customization:")
for sent in doc_custom.sents:
    print("-", sent.text)
```

```python
#4. Beautifulsoup
import requests
from bs4 import BeautifulSoup
from collections import Counter

url = "https://www.theguardian.com/us-news/2022/jun/18/donald-trump-election-defense-fundraising-defrauded-fans"
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')

paragraphs = [p.get_text() for p in soup.find_all('p')]
article_text = " ".join(paragraphs)

doc_article = nlp(article_text)

print("All Named Entities:")
for ent in doc_article.ents:
    print(f"{ent.text} ({ent.label_})")

span_to_visualize = doc_article
displacy.render(span_to_visualize, style="ent", jupyter=True)
```
