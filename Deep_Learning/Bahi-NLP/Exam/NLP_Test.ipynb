```python
# 1. Extract the top 12 most common words in the text ’JunkF’ (on Moodle) excluding stopwords.
# We read the file directly and filter stopwords, punctuation, and spaces.
import spacy
from collections import Counter

nlp = spacy.load("en_core_web_sm")

with open("/JunkFood.rtf", "r", encoding="utf-8") as f:
    text = f.read()

doc = nlp(text)

filtered_words = [token.text.lower() for token in doc if not token.is_stop and not token.is_punct and not token.is_space]

word_freq = Counter(filtered_words)
print("Top 12 words:", word_freq.most_common(12))
```

```python
# 2. Add the custom stopwords ’NIL’ and ’JUNK’ in spaCy and remove the stopwords
# We add custom stopwords to the vocabulary and extract tokens excluding the new stopwords
import spacy

nlp = spacy.load("en_core_web_sm")

nlp.vocab["NIL"].is_stop = True
nlp.vocab["JUNK"].is_stop = True

text = "Jonas was a JUNK great guy NIL Adam was evil NIL Martha JUNK was more of a fool"
doc = nlp(text)

clean_tokens = [token.text for token in doc if not token.is_stop and not token.is_space]
print(clean_tokens)
```

```python
# 3. Extract the usernames from the email addresses
# We use regular expressions to capture the string before the @ symbol
import re

text = "The new registrations are jacques01@gmail.com, elixir101@icloud.com. Please send your work to jacques_bahi@univ-fcomte.fr"

usernames = re.findall(r'([a-zA-Z0-9_.-]+)@', text)
print("Usernames:", usernames)
```

```python
# 4. Identify and extract a list of all organizations

# We filter named entities by the 'ORG' label.
# The en_core_web_sm model is a small statistical model and occasionally misses specific companies like Flipkart.
# For that we add an EntityRuler before the default NER component and then define the exact pattern and label we want to enforce.
import spacy

nlp = spacy.load("en_core_web_sm")

ruler = nlp.add_pipe("entity_ruler", before="ner")

patterns = [
    {"label": "ORG", "pattern": "Flipkart"}
]
ruler.add_patterns(patterns)

text = "Google has released it's new model which has got attention of everyone. Amazon is planning to expand into Food delivery, thereby giving competition Apple is coming up with new iphone model. Flipkart will have to catch up soon."

doc = nlp(text)

organizations = [ent.text for ent in doc.ents if ent.label_ == "ORG"]
print("Organizations:", organizations)
```

```python
# 5. Display the named entities
# We print the entity text and its predefined category label
import spacy

nlp = spacy.load("en_core_web_sm")
text = "Walter was arrested yesterday at Brooklyn for murder. The suspicions and fingerprints pointed to Walter and his friend Pinkman. He is from Paris"

doc = nlp(text)

for ent in doc.ents:
    print(f"{ent.text:12} | {ent.label_}")
```

```python
# 6. Detect all the Laptop names

# "MacBook" is intended to be classified as a PRODUCT, but due to the statistical nature of the models, it is frequently misclassified as an ORG (Organization) or missed entirely.
# Sony and Nvidia are company names, not laptop names.
import spacy
from spacy.matcher import Matcher

nlp = spacy.load("en_core_web_sm")
text = "For my offical use, I prefer MacBook. For gaming purposes, I love Sony and for computing Nvidia is the best"

doc = nlp(text)

laptops = [ent.text for ent in doc.ents if ent.label_ in ["PRODUCT", "ORG"]]
print("Detected Laptops without rule:", laptops)

# Here use spaCy's Matcher or EntityRuler to create rule-based text extraction for specific target vocabulary.
# We initialize the Matcher and then define a pattern for known laptop models (in our case it's just macbook).
# This way we optimise our results and only get Macbook
matcher = Matcher(nlp.vocab)

laptop_patterns = [
    [{"LOWER": "macbook"}]
]

matcher.add("LAPTOP", laptop_patterns)
matches = matcher(doc)

laptops = [doc[start:end].text for match_id, start, end in matches]
print("Detected Laptops with rule:", laptops)
```

**Q7. Give all the approaches to convert texts to vectors and explain the advantages and
advantages of each.**

Answer:


1.   Bag of Words: It represents text by counting word frequencies.
*   Advantages: Simple to implement and computationally cheap.
*   Disadvantages: Ignores word order and context, and creates massive, memory-heavy sparse matrices. Fails to process Out Of Vocabulary words.


2.   Word Embeddings: It uses machine learning to represent words as dense vectors of real numbers.
*   Advantages: Captures semantic and syntactic relationships, placing similar words close to each other in the vector space.
*   Disadvantages: Requires significantly more computational power to train. Traditional embeddings map one vector per word, ignoring multiple meanings of the same word.


3.   Term Frequency Inverse Document Frequency: It weights word frequency against how common the word is across the entire corpus.
*   Advantages: Gives higher weight to rare, distinguishing words specific to a document while penalizing common filler words.
*   Disadvantages: Ignores word order and syntactic relationships.


4.   Contextual Transformers: Pre-trained language models that generate dynamic vectors based on the entire sentence.
*   Advantages: Excellent accuracy because it understands deep, bidirectional context and the specific meaning of a word in a specific sentence.
*   Disadvantages: Extremely resource-intensive to run and train.
