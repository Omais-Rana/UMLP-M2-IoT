```python
import spacy
from collections import Counter

nlp = spacy.load("en_core_web_sm")

with open("/content/completetext.txt", "r", encoding="utf-8") as f:
    complete_text = f.read()

# 1. Create a nlp doc
complete_doc = nlp(complete_text)

# 2. Extract all the tokens
tokens = [token.text for token in complete_doc]
print("2. Tokens")
print(tokens)

# 3. Print all Lemmas
lemmas = [token.lemma_ for token in complete_doc]
print("\n 3. Lemmas")
print(lemmas)

# 4. Remove stop words
filtered_words = [token.text for token in complete_doc if not token.is_stop and not token.is_punct and not token.is_space]
word_freq = Counter(filtered_words)

# 5. Print the 5 commonly occurring words
print("\n 5. Top 5 Common Words")
for word, freq in word_freq.most_common(5):
    print(f"{word}: {freq}")

# 6. Print the unique words
unique_words = set(filtered_words)
print(f"\n 6. Unique Words ({len(unique_words)} total)")
print(list(unique_words))

# 7. POS
print("\n 7. POS")
for token in complete_doc:
    print(f"{token.text:12} | POS: {token.pos_:5} | TAG: {token.tag_:4} | {spacy.explain(token.tag_)}")

# 8. Extract the Nouns and the adjectifs
nouns = [token.text for token in complete_doc if token.pos_ == "NOUN"]
adjectives = [token.text for token in complete_doc if token.pos_ == "ADJ"]

print("\n 8. Nouns and Adjectives")
print("Nouns:", nouns)
print("Adjectives:", adjectives)
```
